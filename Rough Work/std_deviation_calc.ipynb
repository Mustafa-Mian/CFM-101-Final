```python
import yfinance as yf
import pandas as pd

def calculate_stock_std_devs(csv_file):
    # Step 1: Read the list of stock tickers from a CSV file without headers
    tickers_df = pd.read_csv(csv_file, header=None)
    tickers = tickers_df[0].tolist()  # Access the first column directly

    stock_std_devs = []  # List to store (ticker, std_dev) pairs

    # Step 2: Loop through each ticker
    for ticker in tickers:
        try:
            # Fetch historical data
            stock_data = yf.download(ticker, period="1y")  # 1 year of data
            # Calculate daily returns
            stock_data['Daily Return'] = stock_data['Adj Close'].pct_change()
            # Calculate standard deviation of daily returns
            std_dev = stock_data['Daily Return'].std()
            # Append the (ticker, std_dev) pair to the list
            stock_std_devs.append((ticker, std_dev))
        except Exception as e:
            print(f"Could not retrieve data for {ticker}: {e}")

    # Step 3: Sort the list by standard deviation (smallest to largest)
    stock_std_devs.sort(key=lambda x: x[1])

    return stock_std_devs


```

```python
result = calculate_stock_std_devs("Tickers_Example.csv")
print(result)  # Outputs a sorted list of (ticker, std_dev) tuples
```
